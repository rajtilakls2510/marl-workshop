## 3. One frame from each environment family

```python
%matplotlib inline
import matplotlib.pyplot as plt
import gymnasium as gym
import smaclite  # registers the SMAClite envs
from mpe2 import simple_spread_v3, simple_tag_v3, simple_speaker_listener_v4
from vmas import make_env
from marl.utils import vmas_frame

frames = {}
for name, mod in [('MPE simple_spread', simple_spread_v3), ('MPE simple_tag', simple_tag_v3),
                  ('MPE speaker_listener', simple_speaker_listener_v4)]:
    env = mod.parallel_env(render_mode='rgb_array', max_cycles=25)
    env.reset(seed=0)
    for _ in range(5):
        env.step({a: env.action_space(a).sample() for a in env.agents})
    frames[name] = env.render(); env.close()

senv = gym.make('smaclite/2s3z-v0', render_mode='rgb_array'); senv.reset(seed=0)
frames['SMAClite 2s3z'] = senv.render(); senv.close()

venv = make_env('navigation', num_envs=8, device='cpu', continuous_actions=True); venv.reset()
venv.step([venv.get_random_action(a) for a in venv.agents])
frames['VMAS navigation'] = vmas_frame(venv)  # (VMAS's first render() is blank; the helper handles it)

fig, axes = plt.subplots(1, len(frames), figsize=(4 * len(frames), 4))
for ax, (title, fr) in zip(axes, frames.items()):
    ax.imshow(fr); ax.set_title(title, fontsize=10); ax.axis('off')
plt.tight_layout(); plt.show()
```
